## 1. Define Problem

### Bài toán

Dự đoán chi phí bảo hiểm y tế (charges) dựa trên thông tin cá nhân và đặc điểm sức khỏe:

$X = \{ age, sex, bmi, children, smoker, region\}$

$y = charges$

Đây là bào toán: `Supervised Learning + Regression`

```
Input
 ├── age
 ├── sex
 ├── bmi
 ├── children
 ├── smoker
 └── region
       ↓
      ANN
       ↓
Predicted charges
```

tìm hàm sấp xỉ.

4. Feature nào categorical?

6. Có missing values không?
7. Có duplicate rows không?
8. Numerical features có phân phối như thế nào?
9. Có outlier không?
10. Categorical features có bao nhiêu category?
11. Các nhóm categorical khác nhau ảnh hưởng đến target như thế nào?
12. Target `charges` có phân phối như thế nào?
13. Feature nào có mối quan hệ đáng chú ý với target?
